### 2. Dataset opschonen en verkennen

```python
# Dataframe buurten aanmaken: alleen rijen waar SoortRegio 'Buurt' is
# .copy() maakt er een los dataframe van, zodat we later kolommen kunnen toevoegen
buurten = cbsdata[cbsdata['SoortRegio_2'].str.strip() == 'Buurt'].copy()
buurten.head()
```

#### Waarom is het nodig om alleen op buurten te filteren en niet alle soorten erin te laten staan?
We filteren op alleen buurten omdat de dataset verschillende niveaus door elkaar heeft: Nederland, gemeenten, wijken en buurten. Als je die allemaal laat staan, tel je mensen dubbel, want een inwoner zit ook in zijn wijk, gemeente en Nederland. Daarnaast kun je een hele gemeente niet eerlijk vergelijken met één kleine buurt. Door alleen buurten te houden zit alles op hetzelfde niveau (dezelfde granulariteit) en is elke rij één buurt.

#### Onderbouwing te veel missende waardes
We verwijderen een kolom als *40% of meer* van de waarden mist.

- Kuhn en Johnson (2019) adviseren om eerst kolommen met veel missende waarden te verwijderen, en niet de buurten zelf. Zo houd je zoveel mogelijk buurten over. Kolommen waarvan je weet dat ze belangrijk zijn, verwijder je niet.
- Hvitfeldt (z.d.) noemt als grens 30% of 50% missende waarden. Wij kiezen 40%, daar tussenin. Bij 30% zouden we ook opleiding en het verbruik per woningtype verwijderen. Die missen 34 tot 40%, vooral omdat het CBS cijfers van kleine buurten geheimhoudt, maar ze zijn wel relevant voor het energieverbruik.

In onze data valt alleen stadsverwarming boven de grens (92,5% mist). De andere kolommen missen 39,9% of minder.

Bronnen:
- Kuhn, M., & Johnson, K. (2019). Feature Engineering and Selection: A Practical Approach for Predictive Models (hoofdstuk 8.3). CRC Press. https://bookdown.org/max/FES/deletion-of-data.html
- Hvitfeldt, E. (z.d.). Feature Engineering A-Z (hoofdstuk Remove Missing Values). https://feaz-book.com/missing-remove

```python
# Toont het percentage missende waarden per kolom, hoogste bovenaan
percentage_missend = buurten.isna().mean() * 100
percentage_missend.sort_values(ascending=False).head()
```

```python
# Verwijderen van kolommen met te veel missende waardes (40% of meer)
te_veel_missend = percentage_missend[percentage_missend >= 40].index
buurten = buurten.drop(columns=te_veel_missend)
print('Verwijderd:', list(te_veel_missend))
print('Over:', buurten.shape)
```

### 3. Dataset beschrijven met basisstatistiek
#### Totalen per gemeente
We tellen de buurten per gemeente op. Dat kan alleen bij kolommen met aantallen, niet bij percentages of gemiddelden.

```python
# Basisstatistieken:
# Totalen per gemeente

# Alle overgebleven kolommen met aantallen (de rest zijn percentages, gemiddelden of codes)
aantal_kolommen = ['AantalInwoners_5', 'Mannen_6', 'Vrouwen_7', 'k_0Tot15Jaar_8', 'k_15Tot25Jaar_9',
                   'k_25Tot45Jaar_10', 'k_45Tot65Jaar_11', 'k_65JaarOfOuder_12', 'HuishoudensTotaal_29',
                   'Eenpersoonshuishoudens_30', 'HuishoudensZonderKinderen_31', 'HuishoudensMetKinderen_32',
                   'Woningvoorraad_35', 'BasisonderwijsVmboMbo1_70', 'HavoVwoMbo24_71', 'HboWo_72',
                   'AantalInkomensontvangers_79', 'PersonenautoSTotaal_109', 'PersonenautoSBrandstofBenzine_110',
                   'PersonenautoSOverigeBrandstof_111', 'Motorfietsen_114']

totalen_gemeente = buurten.groupby('Gemeentenaam_1')[aantal_kolommen].sum()
totalen_gemeente.head()
```

```python
# Vergelijken met het CBS per gemeente
# CBS-cijfers per gemeente: de gemeenterijen uit de dataset (dezelfde cijfers als op de CBS-website)
gemeenten = cbsdata[cbsdata['SoortRegio_2'].str.strip() == 'Gemeente']
gemeenten = gemeenten.set_index(gemeenten['Gemeentenaam_1'].str.strip())[aantal_kolommen]

# Spaties achter de gemeentenamen weghalen, zodat onze namen gelijk zijn aan die van het CBS
totalen_gemeente.index = totalen_gemeente.index.str.strip()

# Per kolom drie kolommen naast elkaar: onze som van de buurten, het CBS-cijfer en het verschil
vergelijking = pd.concat({'som buurten': totalen_gemeente,
                          'CBS': gemeenten,
                          'verschil': totalen_gemeente - gemeenten}, axis=1)
vergelijking = vergelijking.swaplevel(axis=1)[aantal_kolommen]
vergelijking
```

#### Verschillen met het CBS
We vergelijken per gemeente onze som van de buurten met het cijfer van het CBS. Dit zijn de gemeenterijen uit de dataset, dezelfde cijfers als op de CBS-website (StatLine, tabel 85618NED). Bijvoorbeeld 's-Gravenhage: 562.839 inwoners volgens het CBS, 562.490 als we de buurten optellen.

- *Inwoners, leeftijd, huishoudens en inkomensontvangers:* bij bijna alle gemeenten een klein verschil. Bij inwoners is het verschil meestal minder dan 15, het grootste is 349 (Den Haag). Het CBS rondt deze aantallen per buurt af op vijftallen vanwege privacy. Optellen van afgeronde getallen geeft een iets ander totaal.
- *Woningvoorraad:* bij alle gemeenten precies gelijk. Die wordt niet afgerond.
- *Opleidingsniveau:* bij bijna alle gemeenten lager dan het CBS. Bij ongeveer een derde van de buurten is het opleidingsniveau geheim (NaN). Die buurten tellen bij ons niet mee, bij het CBS wel.
- *Auto's:* bij de meeste gemeenten bijna gelijk, maar bij een paar gemeenten veel lager, zoals Houten, Almere en Haarlemmermeer. Waarschijnlijk staan daar veel auto's op naam van bedrijven. Die telt het CBS wel bij de gemeente, maar kan het niet aan een buurt koppelen. Dit hebben we niet kunnen controleren.

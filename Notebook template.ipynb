```python
# run dit in terminal: pip install cbsodata
# Dataset importeren
import cbsodata 
cbsdata = pd.DataFrame(cbsodata.get_data('85618NED'))
cbsdata.shape
```

```python
# Dataframe buurten maken
buurten = cbsdata[cbsdata['SoortRegio_2'].str.strip() == 'Buurt']
buurten.head()
```

#### Waarom is het nodig om alleen op buurten te filteren en niet alle soorten erin te laten staan?
We filteren op alleen buurten omdat de dataset verschillende niveaus door elkaar heeft: Nederland, gemeenten, wijken en buurten. Als je die allemaal laat staan, tel je mensen dubbel, want een inwoner zit ook in zijn wijk, gemeente en Nederland. Daarnaast kun je een hele gemeente niet eerlijk vergelijken met één kleine buurt. Door alleen buurten te houden zit alles op hetzelfde niveau (dezelfde granulariteit) en is elke rij één buurt. 



```python
# Aantal rijen en kolommen dataframe
print(buurten.shape)
```

```python
# code die gebruikt is
overzicht = pd.DataFrame({
    "kolomnaam": buurten.columns[0:32],
    "datatype": buurten.dtypes[0:32].astype(str),
    "waarden": buurten[buurten.columns[0:32]].count(),
    "missende waarden": buurten[buurten.columns[0:32]].isna().sum()
})

overzicht

mijn_num = [kolom for kolom in buurten.columns[0:32]
            if pd.api.types.is_numeric_dtype(buurten[kolom]) and kolom != "ID"]

buurten[mijn_num].agg(["max", "min", "mean", "median"]).T

```
